```python

import pandas as pd

# Read in the two CSV files
df1 = pd.read_csv('DC_full copy.csv')
df2 = pd.read_csv('Test-out.csv')

# Compare the dataframes (this requires pandas 1.1.0 or later)
diff = df1.compare(df2, keep_shape=True, keep_equal=False)

# Filter out rows where there are differences (rows with at least one difference)
diff_rows = diff[diff.isnull().any(axis=1)]  # Rows where there's at least one difference

# Create an HTML file to display the differences
html_output = diff_rows.style.applymap(lambda x: 'background-color: yellow' if pd.notnull(x) else '')

# Save the HTML output to a file
html_output.to_html('highlighted_diff_output.html')

# Output confirmation message
print("Differences highlighted and saved to 'highlighted_diff_output.html'.")
```
