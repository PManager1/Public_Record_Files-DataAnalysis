```python
import pandas as pd

# Read the main data file
df = pd.read_csv('DC_full_copy.csv')

# Specify the columns to search and move data from
search_column = 'Grantee'  # Column to search for 'LLC' or 'Inc'
move_from_column = 'Doc Type'  # Column from which the value will be moved
move_to_column = 'Grantee'  # Column where the value will be moved to

# Find rows where 'LLC' or 'Inc' is in the 'Grantee' column
mask = df[search_column].str.contains('LLC|Inc|INC.|N.A.|L.P.|L.L.C.|Inc|LTD.|LTD', case=False, na=False)

# Create a copy of the rows that will be changed (before modification)
changed_records = df[mask].copy()

# Move the value from 'Doc Type' to 'Grantee' column for these rows
df.loc[mask, move_to_column] = df.loc[mask, move_from_column]

# Save the updated DataFrame to a new CSV file
df.to_csv('Test-out.csv', index=False)

# Save the changed records (before modification) to a separate CSV file
changed_records.to_csv('ChangedRecords.csv', index=False)

# Display confirmation message
print("Updated data saved to 'Test-out.csv'. Changed records saved to 'ChangedRecords.csv'.")
```
