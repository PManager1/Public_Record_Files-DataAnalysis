```python
import pandas as pd

# Read the excel file
dfk = pd.read_csv('KeywordsRem.csv') 
# dfk = pd.read_csv('KeywordsRemNew.csv')  

# Get the list of values
caps = dfk['Keywords'].tolist()
lowers = [x.lower() for x in caps]
elimkeys = caps + lowers

# Read the main data file
df = pd.read_csv('DC_full.csv', encoding="ISO-8859-1" , low_memory=False)  # or encoding="latin1"

# Filter out rows containing keywords
df = df[~df['Grantee'].str.contains('|'.join(elimkeys), na=False)]

# Get value counts
r = df['Grantee'].value_counts()

# Save results to CSV
r.to_csv('output-dc.csv', index=True)

print("Outputted results saved to 'output-dc-2.csv'")
```
